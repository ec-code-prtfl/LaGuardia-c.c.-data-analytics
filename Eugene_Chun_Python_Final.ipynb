## *****Creating missing data**:

```python
df.loc[df['experience_years'].sample(frac=0.03).index,'experience_years']=np.nan
df.loc[df['skills_count'].sample(frac=0.01).index,'skills_count']=np.nan
df.loc[df['certifications'].sample(frac=0.02).index,'certifications']=np.nan
df.loc[df['salary'].sample(frac=0.009).index,'salary']=np.nan
df.loc[df['company_size'].sample(frac=0.00007).index,'company_size']=np.nan

```

```python
df.isna().sum()
```

```python

x=np.array(['experience_years','skills_count','certifications','salary','company_size'])
y=np.array([20000,5000,10000,5000,1000])
plt.figure(figsize=(10,5))
plt.bar(x,y,width=.5,color='#4CAF50')
plt.title("Visualization of missing data of categorical columns")
plt.ylabel('Count of isna')
plt.xlabel('Categorical Columns')
plt.show()
```

```python
print('\nData conversion to numeric')

df['experience_years']=pd.to_numeric(df['experience_years'],errors='coerce')
df['skills_count']=pd.to_numeric(df['skills_count'],errors='coerce')
df['certifications']=pd.to_numeric(df['certifications'],errors='coerce')
df['salary']=pd.to_numeric(df['salary'],errors='coerce')
df['company_size']=pd.to_numeric(df['salary'],errors='coerce')
```

```python
df['experience_years']=df['experience_years'].astype(float).fillna(df['experience_years'].astype(float).median())
df['skills_count']=df['skills_count'].astype(float).fillna(df['skills_count'].astype(float).median())
df['certifications']=df['certifications'].astype(float).fillna(df['certifications'].astype(float).median())
df['salary']=df['salary'].astype(float).fillna(df['salary'].astype(float).mean())
df['company_size']=df['company_size'].astype(float).fillna(df['salary'].astype(float).median())


print('Numerical imputation')
```

```python
print('Verify data set is clean')
df.isna()
```

```python
#re-validate dataset is clean
df.isna().sum()
```

```python
categorical_cols=['experience_years','skills_count','company_size','location','certifications','salary'
]
for col in categorical_cols:
  df[col]=df[col].astype('category')
df[categorical_cols].dtypes
```

```python
def show_category_distribution(col_name):
  counts=df[col_name].valuecounts()
  percents=df[col_name].value_counts(normalize=True)*100
  display(pd.DataFrame({'count':counts,'percent':percents.round(2)}))
  for col in ['job_title','experience_years','education_level','skills_count','industry','company_size','location','remote_work','certifications','salary'
]:
    print(f'---{col}---')
    show_category_distribution(col)
    print()
```

Section 5: Correlation Analysis

```python
#correlation matrix

corr_matrix=df.select_dtypes(include=['int64','float64']).corr()
corr_matrix
```

The corrrelation matrix, Heatmap, reveals that experience is a significant factor in relation to salary (0.43). Skills matter somewhat (0.13). Certifications has the lowest relationship with salary at 0.074.
The -0.00 indicates that more experience does not necessarily mean more certifications or skills. We can assume these variable are independent.  



```python
plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix,annot=True,fmt='.2f',cmap='coolwarm',linewidths=0.5)
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```
